NumPy Learning 

```python
import numpy as np
```

Step 1: To Generate Data

```python

# 50 Students, 5 Subjects
# Shape: (50, 5)
np.random.seed(42) # Taake har baar same random numbers aayein
grades = np.random.randint(0, 101, (50, 5))

print("Grades Array Shape:", grades.shape)
```

Step 2: To find Statistics (Axis Logic)

```python
# 1. Har student ke average marks
student_averages = np.mean(grades, axis=1)

# 2. Har subject ka class average
subject_averages = np.mean(grades, axis=0)

print(f"Pehle student ka average: {student_averages[0]:.2f}")
print(f"Har subject ka average: {subject_averages}")
```

Step 3: Filtering and Masking (Important Concept)

```python
# Boolean Masking
fail_filter = student_averages < 40
failed_students_indices = np.where(fail_filter)[0]

print(f"Total students jo fail huay: {len(failed_students_indices)}")
print(f"Fail hone wale students ke Roll Numbers (Indices): {failed_students_indices}")
```

Step 4: Find Class Topper

```python
overall_topper_index = np.argmax(student_averages)
print(f"Class Topper Roll No: {overall_topper_index} with Average: {student_averages[overall_topper_index]:.2f}")
```
